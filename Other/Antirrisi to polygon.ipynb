### Run as is, to import libraries

```python
import arcpy
import os
import re
```

### Use a desired name. A Feature class will be created on the project's geodatabase

```python
dataset_name = "Antirrisi_Pappadaki"
```

### Run below as is

```python
aprx = arcpy.mp.ArcGISProject("CURRENT")  # or path to .aprx file
default_gdb = aprx.defaultGeodatabase
if not arcpy.Exists(os.path.join(default_gdb,dataset_name)):
    arcpy.management.CreateFeatureclass(
        out_path=aprx.defaultGeodatabase,
        out_name=dataset_name,
        geometry_type="POLYGON",
        template=None,
        has_m="DISABLED",
        has_z="DISABLED",
        spatial_reference=arcpy.SpatialReference(2100),
        config_keyword="",
        spatial_grid_1=0,
        spatial_grid_2=0,
        spatial_grid_3=0,
        out_alias="",
        oid_type="SAME_AS_TEMPLATE"
    )
    
    arcpy.management.AddFields(
        in_table=os.path.join(aprx.defaultGeodatabase,dataset_name),
        field_description="aa_old SHORT 'Α.Α. ΑΡΧΙΚΟ' # # # # # # #;aa_new SHORT 'Α.Α. ΝΕΟ' # # # # # # #;emvadon DOUBLE # # # # # # # #;new_proteuon TEXT 'Νέος Πρωτεύων' # # # # # true #;new_deutereuon TEXT 'Νέος Δευτερεύων' # # # # # true #;new_triteuon TEXT 'Νέος Τριτεύων' # # # # # true #;new_landtype TEXT 'Νέο Landtype' # # # # # # #",
        template=None
    )

def insert_antirrisi(fc_path, syntetagmenes, aa_old, aa_new, emvadon,
                     new_proteuon, new_deutereuon, new_triteuon, new_landtype):
    sr = arcpy.SpatialReference(2100)

    points = []
    for match in re.finditer(r'X\d+=(\d+),(\d+)\s+Y\d+=(\d+),(\d+)', syntetagmenes):
        x = float(f"{match.group(1)}.{match.group(2)}")
        y = float(f"{match.group(3)}.{match.group(4)}")
        points.append(arcpy.Point(x, y))

    polygon = arcpy.Polygon(arcpy.Array(points), sr)

    fields = [
        "SHAPE@",
        "aa_old", "aa_new", "emvadon",
        "new_proteuon", "new_deutereuon", "new_triteuon", "new_landtype"
    ]

    with arcpy.da.InsertCursor(fc_path, fields) as cursor:
        cursor.insertRow([
            polygon,
            aa_old, aa_new, emvadon,
            new_proteuon, new_deutereuon, new_triteuon, new_landtype
        ])
```

### Run below with data from pdf

```python
aa_old=1
aa_new=1001
emvadon=17.48
syntetagmenes="""
X1=632352,223 Y1=4106490,677
X2=632340,665 Y2=4106481,786
X3=632341,078 Y3=4106485,129
X4=632352,223 Y4=4106490,677
"""
new_proteuon='ΠΔ'
new_deutereuon=''
new_triteuon=''
new_landtype='0'
##--##
insert_antirrisi(
    fc_path=os.path.join(default_gdb, dataset_name),
    syntetagmenes=syntetagmenes,
    aa_old=aa_old,
    aa_new=aa_new,
    emvadon=emvadon,
    new_proteuon=new_proteuon,
    new_deutereuon=new_deutereuon,
    new_triteuon=new_triteuon,
    new_landtype=new_landtype
)
```
